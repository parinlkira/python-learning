```python
#string def 1 using ''
s1 = 'Hello'

#def2 using double quotes
s2 = "Hello Double"

#def3 using 3 single quotes
s3 = '''Hello Triple Quotes'''

#def4 using 3 double quotes helps break into multiline comment
s4 = """Hello
How are you?
"""

# Enclose in "" quotes if you want to write an ' like Clark's
s5 = "Clark's"

# Enclose in '' quotes if you want to write an " like Clarks"s
s6 = 'Clark"s'

print(s1, s2, s3, s4, s5, s6)
```

```python
s1 = 'Hello' 
# The way to access strings is below like arrays, but in python
# you call arrays as sequence, so string is a sequence of characters starting from
# 0th index just like java
print(s1[0]) 

#Only in python you can have negative indexes as well, starting from -1 from right
#i.e. the end of the string and going -2 and so on
print(s1[-1])
```

```python
#using the len() function to calculate the length of a string
print(len(s1))
```

```python
#string traversal

for x in s1:
    print(x)

#way 2 
for i in range(len(s1)):
    print(s1[i])
```
