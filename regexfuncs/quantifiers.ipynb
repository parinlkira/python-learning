```python
import re

re.fullmatch('(ab)?', 'ab') # Returns a match object if 'ab' is present zero or one time
```

```python
re.fullmatch('(ab)?', '') # Returns a match object if 'ab' is present zero or one time
```

```python
re.fullmatch('(ab)*','') # Returns a match object if 'ab' is present zero or more times
```

```python
re.fullmatch('(ab)+', 'ababab') # Returns a match object if 'ab' 
# is present one or more times
```

```python
re.findall('[abc]+', '123 abc 987 bbcc@cbac') # Returns all occurrences of one 
# or more consecutive 'a', 'b', or 'c' characters
```

```python
re.fullmatch('(ab){2}', 'abab') # Returns a match object if 'ab' is present exactly two times
re.fullmatch('(ab){2,4}', 'ababab') # Returns a match object if 'ab' is present between two and four times
re.fullmatch('(ab){2,4}', 'ab') # Returns None because 'ab' is present less than two times
re.fullmatch('(ab){2,4}', 'abababab') # Returns None if 'ab' is present more than four times
```
