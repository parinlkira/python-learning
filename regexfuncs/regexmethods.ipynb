```python
#import regex has three ways 
# 1. import re
# 2. from re import *
# 3. import re as regex : here you can use 'regex' instead of 're' 
# to access regex functions

import re
```

```python
re.match('abc', 'abcdef')   
```

```python
re.fullmatch('python','python').group() # Returns the entire match if the whole string matches the pattern
```

```python
re.search('very', 'python is very easy').span()  # Returns the start and end positions of the match
```

```python
re.findall('can', 'can you can a can as a canner')
```
